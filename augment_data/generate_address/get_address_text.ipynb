```python
import os

def save_lists_to_txt(data, output_dir):
    """Save each list to a separate .txt file"""
    os.makedirs(output_dir, exist_ok=True)

    for key, values in data.items():
        file_path = os.path.join(output_dir, f"{key}.txt")
        with open(file_path, "w", encoding="utf-8") as f:
            for item in values:
                f.write(item + "\n")
        print(f"✅ Saved {len(values)} items to {file_path}")


# Example usage
# -------------------------------------------------
# path_to_input = "path/to/your/file.txt"
# output_dir = "vietnam_admin_with_prefix"
# data = extract_admin_units(path_to_input)
# save_lists_to_txt(data, output_dir)
```

```python
import os

COMMUNE_PREFIXES = ["Xã", "Phường", "Thị trấn"]
DISTRICT_PREFIXES = ["Huyện", "Quận", "Thị xã", "Thành phố"]
PROVINCE_PREFIXES = ["Tỉnh", "Thành phố"]
```

```python
def remove_prefix(name, prefixes):
    """Xóa tất cả prefix khỏi chuỗi (vd: 'Xã A, Huyện B, Tỉnh C' -> 'A, B, C')"""
    parts = [p.strip() for p in name.split(",")]
    cleaned_parts = []

    for part in parts:
        part = part.strip()
        for prefix in prefixes:
            if part.startswith(prefix + " "):
                part = part[len(prefix) + 1:].strip()
                break
        cleaned_parts.append(part)

    return ", ".join(cleaned_parts)
```

```python
def create_no_prefix_versions(input_dir):
    """Đọc các file .txt, tạo bản no prefix"""
    for filename in os.listdir(input_dir):
        if not filename.endswith("_with_prefix.txt"):
            continue

        file_path = os.path.join(input_dir, filename)
        no_prefix_file = os.path.join(
            input_dir,
            filename.replace(".txt", "_no_prefix.txt")
        )

        if "commune" in filename:
            prefixes = COMMUNE_PREFIXES
        elif "district" in filename:
            prefixes = DISTRICT_PREFIXES
        elif "province" in filename:
            prefixes = PROVINCE_PREFIXES
        else:
            prefixes = COMMUNE_PREFIXES + DISTRICT_PREFIXES + PROVINCE_PREFIXES  # fallback

        with open(file_path, "r", encoding="utf-8") as f:
            lines = [line.strip() for line in f if line.strip()]

        cleaned = [remove_prefix(line, prefixes) for line in lines]

        with open(no_prefix_file, "w", encoding="utf-8") as f:
            for item in cleaned:
                f.write(item + "\n")

        print(f"✅ Created no-prefix version: {no_prefix_file} ({len(cleaned)} lines)")
```

```python
input_dir = "/media/nampv1/hdd/data/vn_commune_district_province/raw/text/"
create_no_prefix_versions(input_dir)
```
