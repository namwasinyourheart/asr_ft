```python
from huggingface_hub import hf_hub_download, list_repo_files
from tqdm import tqdm
import os

# Local directory to store the dataset
LOCAL_DIR = "/media/nampv1/hdd/data/ASR-GigaSpeech2-Vi/raw/"

# Hugging Face repo containing the dataset
REPO_ID = "speechcolab/gigaspeech2"

def list_all_files():
    """List all files inside data/vi/ (including dev, test, train...)."""
    all_files = list_repo_files(repo_id=REPO_ID, repo_type="dataset")
    return [f for f in all_files if f.startswith("data/vi/")]

def download_files(files):
    """Download all files in the given list with progress bar."""
    for file_path in tqdm(files, desc="Downloading files", unit="file"):
        hf_hub_download(
            repo_id=REPO_ID,
            filename=file_path,
            repo_type="dataset",
            local_dir=LOCAL_DIR,
            local_dir_use_symlinks=False,
        )

if __name__ == "__main__":
    files_to_download = list_all_files()
    print(f"Total number of files to download: {len(files_to_download)}")
    download_files(files_to_download)
```
