```python
def load_districts_from_file(file_path):
    """
    Load tên quận/huyện từ file .txt (tab-separated) và loại bỏ
    các từ Quận/Huyện/Thị xã ở đầu tên.
    """
    remove_prefixes = ["Quận", "Huyện", "Thị xã"]
    districts = []

    with open(file_path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    for line in lines[1:]:  # bỏ header
        parts = line.strip().split("\t")
        if len(parts) >= 2:
            name = parts[1].strip()
            # Loại bỏ prefix nếu có
            for prefix in remove_prefixes:
                if name.startswith(prefix + " "):
                    name = name[len(prefix)+1:]
                    break
            if name:
                districts.append(name)
    return districts

# --- Demo ---
district_list = load_districts_from_file(old_districts_696_not_processed_path)
print(district_list)
print(len(district_list))
```

```python
def save_districts_to_file(district_list, output_file):
    """
    Lưu danh sách districts ra file, mỗi dòng một tên.
    """
    with open(output_file, "w", encoding="utf-8") as f:
        for district in district_list:
            f.write(district + "\n")
    print(f"Saved {len(district_list)} districts to {output_file}")

# --- Demo sử dụng ---
district_list = load_districts_from_file(old_districts_696_not_processed_path)
save_districts_to_file(district_list, "/home/nampv1/projects/asr/asr-evaluation-app/test_samples/old_districts_list_696.txt")
```

```python
def load_wards_from_file(file_path):
    """
    Load tên xã/phường/thị trấn từ file .txt (tab-separated)
    và loại bỏ tiền tố Xã / Phường / Thị trấn ở đầu tên.
    """
    remove_prefixes = ["Xã", "Phường", "Thị trấn"]
    wards = []

    with open(file_path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    for line in lines[1:]:  # bỏ header
        parts = line.strip().split("\t")
        if len(parts) >= 1:
            name = parts[0].strip()  # cột 0 chứa tên xã/phường
            # Loại bỏ prefix nếu có
            for prefix in remove_prefixes:
                if name.startswith(prefix + " "):
                    name = name[len(prefix)+1:]
                    break
            if name:
                wards.append(name)
    return wards

def save_wards_to_file(ward_list, output_file):
    """
    Lưu danh sách wards ra file mới, mỗi dòng một tên
    """
    with open(output_file, "w", encoding="utf-8") as f:
        for ward in ward_list:
            f.write(ward + "\n")
    print(f"Saved {len(ward_list)} wards to {output_file}")
```

```python
# --- Demo ---
old_communes_10047_not_processed_path = "/home/nampv1/projects/asr/asr-evaluation-app/test_samples/old_communes_10047_not_processed.txt"
commune_list = load_wards_from_file(old_communes_10047_not_processed_path)
save_wards_to_file(commune_list, "/home/nampv1/projects/asr/asr-evaluation-app/test_samples/old_communes_list_10047.txt")
```
