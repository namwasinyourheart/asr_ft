```python
# Cập nhật: cho phép tùy chọn cách đọc chữ "4" ở hàng đơn vị:
# - four_read="auto" (mặc định): đọc "tư" khi hàng chục >= 2, ngược lại "bốn"
# - four_read="tu": luôn đọc "tư" ở hàng đơn vị
# - four_read="bon": luôn đọc "bốn" ở hàng đơn vị
def number_to_vietnamese(n, zero_read="lẻ", four_read="auto"):
    units = ["", "một", "hai", "ba", "bốn", "năm", "sáu", "bảy", "tám", "chín"]
    tens_words = ["", "mười", "hai mươi", "ba mươi", "bốn mươi", "năm mươi",
                  "sáu mươi", "bảy mươi", "tám mươi", "chín mươi"]
    scales = ["", "nghìn", "triệu", "tỷ", "nghìn tỷ", "triệu tỷ", "tỷ tỷ"]

    def read_three_digits(num, is_first_group=False):
        hundred = num // 100
        ten = (num % 100) // 10
        one = num % 10
        parts = []

        # Hàng trăm
        if hundred > 0:
            parts.append(units[hundred] + " trăm")
        elif not is_first_group and num != 0:
            parts.append("không trăm")

        # Hàng chục
        if ten > 1:
            parts.append(tens_words[ten])
        elif ten == 1:
            parts.append("mười")
        else:  # ten == 0
            if one > 0 and (hundred > 0 or not is_first_group):
                parts.append(zero_read)

        # Hàng đơn vị
        if one > 0:
            # chọn đọc "4" theo chế độ four_read
            def read_four(current_ten):
                if four_read == "auto":
                    return "tư" if current_ten >= 2 else "bốn"
                elif four_read == "tu":
                    return "tư"
                else:
                    return "bốn"

            if ten == 0 or ten == 1:
                if one == 5 and ten > 0:
                    parts.append("lăm")
                else:
                    # mặc định lấy từ units (bốn) cho trường hợp ten==0 hoặc ten==1,
                    # nhưng nếu user yêu cầu four_read="tu", cho phép đọc "tư"
                    if one == 4:
                        parts.append(read_four(ten))
                    else:
                        parts.append(units[one])
            else:
                # ten >= 2: xử lý mốt/tư/lăm theo quy tắc
                if one == 1:
                    parts.append("mốt")
                elif one == 4:
                    parts.append(read_four(ten))
                elif one == 5:
                    parts.append("lăm")
                else:
                    parts.append(units[one])

        return " ".join(parts)

    if n == 0:
        return "không"

    s = str(n)
    groups = []
    while s:
        groups.insert(0, int(s[-3:]))
        s = s[:-3]

    words = []
    group_len = len(groups)

    for idx, g in enumerate(groups):
        is_first_group = (idx == 0)
        if g != 0:
            group_words = read_three_digits(g, is_first_group)
            if group_words:
                words.append(group_words)
                scale = scales[group_len - idx - 1]
                if scale:
                    words.append(scale)
    return " ".join(words).strip()


# ---------- Test cases cho việc đọc "4" (auto/tu/bon) ----------
def run_four_tests():
    cases = [
        # đơn vị
        (4, "bốn", "tư", "bốn"),          # 4: auto->bốn, tu->tư, bon->bốn
        (14, "mười bốn", "mười bốn", "mười bốn"),  # 14: luôn 'bốn' (ten==1)
        (24, "hai mươi tư", "hai mươi tư", "hai mươi bốn"),  # 24: tens>=2 -> auto tư; bon forces bốn

        # nhóm 3 chữ số
        (104, "một trăm lẻ bốn", "một trăm lẻ tư", "một trăm lẻ bốn"),
        (304, "ba trăm lẻ bốn", "ba trăm lẻ tư", "ba trăm lẻ bốn"),
        (324, "ba trăm hai mươi tư", "ba trăm hai mươi tư", "ba trăm hai mươi bốn"),

        # nghìn / triệu
        (1004, "một nghìn không trăm lẻ bốn", "một nghìn không trăm lẻ tư", "một nghìn không trăm lẻ bốn"),
        (1204, "một nghìn hai trăm lẻ bốn", "một nghìn hai trăm lẻ tư", "một nghìn hai trăm lẻ bốn"),
        (678934, "sáu trăm bảy mươi tám nghìn chín trăm ba mươi tư",  # auto tư
         "sáu trăm bảy mươi tám nghìn chín trăm ba mươi tư",
         "sáu trăm bảy mươi tám nghìn chín trăm ba mươi bốn"),

        # lớn hơn
        (1000004, "một triệu không trăm lẻ bốn", "một triệu không trăm lẻ tư", "một triệu không trăm lẻ bốn"),
    ]

    for item in cases:
        num = item[0]
        expect_auto = item[1]
        expect_tu = item[2]
        expect_bon = item[3]
        got_auto = number_to_vietnamese(num, zero_read="lẻ", four_read="auto")
        got_tu = number_to_vietnamese(num, zero_read="lẻ", four_read="tu")
        got_bon = number_to_vietnamese(num, zero_read="lẻ", four_read="bon")
        print(f"{num} -> auto: {got_auto}  | expected: {expect_auto}  | {'PASS' if got_auto==expect_auto else 'FAIL'}")
        print(f"{num} -> tu:   {got_tu}  | expected: {expect_tu}  | {'PASS' if got_tu==expect_tu else 'FAIL'}")
        print(f"{num} -> bon:  {got_bon}  | expected: {expect_bon}  | {'PASS' if got_bon==expect_bon else 'FAIL'}")
        print()

run_four_tests()
```

```python
# Chuyển chuỗi đọc tiếng Việt thành số nguyên
# Hỗ trợ các biến thể: "lẻ"/"linh", "mốt"/"một", "tư"/"bốn", "lăm"/"năm"
# Và các lớp: "nghìn", "triệu", "tỷ"
def vietnamese_to_number(text):
    text = text.strip().lower()
    if text == "" or text == "không":
        return 0

    units_map = {
        "không": 0, "một": 1, "mốt": 1, "hai": 2, "ba": 3,
        "bốn": 4, "tư": 4, "năm": 5, "lăm": 5, "sáu": 6,
        "bảy": 7, "tám": 8, "chín": 9
    }
    scales_map = {
        "tỷ": 10**9,
        "triệu": 10**6,
        "nghìn": 10**3,
        "ngàn": 10**3,  # một số người viết "ngàn"
    }

    tokens = text.split()
    total = 0
    group = 0  # giá trị của nhóm 3 chữ số hiện tại (0..999)
    i = 0
    L = len(tokens)

    while i < L:
        t = tokens[i]

        # Hàng trăm: "<digit> trăm" hoặc "không trăm"
        if t in units_map and i + 1 < L and tokens[i + 1] == "trăm":
            group += units_map[t] * 100
            i += 2
            continue
        if t == "không" and i + 1 < L and tokens[i + 1] == "trăm":
            # "không trăm" -> nhóm trăm = 0 (không cộng), nhưng giữ để xử lý tiếp
            # chỉ cần nhảy qua
            i += 2
            continue

        # Hàng chục: "<digit> mươi" (ví dụ "hai mươi")
        if t in units_map and i + 1 < L and tokens[i + 1] == "mươi":
            group += units_map[t] * 10
            i += 2
            continue

        # "mười" (10..19)
        if t == "mười":
            group += 10
            i += 1
            continue

        # Từ "mươi" nếu xảy ra đơn lẻ (hiếm) -> bỏ qua
        if t == "mươi":
            i += 1
            continue

        # Từ lẻ/linh: chỉ dấu tens = 0, không cộng gì
        if t in ("lẻ", "linh"):
            i += 1
            continue

        # Các số đơn vị (một, hai, mốt, tư, lăm, ...)
        if t in units_map:
            group += units_map[t]
            i += 1
            continue

        # Tên lớp (scale): nhân nhóm hiện tại rồi cộng vào tổng
        if t in scales_map:
            scale_val = scales_map[t]
            # Nếu group == 0 thì nhóm mặc định là 1? KHÔNG: trong tiếng VN,
            # thường phải có nhóm trước scale (ví dụ "một tỷ ..."), nhưng
            # nếu viết "tỷ" một mình thì ta không biết; ở đây chỉ xử lý group hiện có.
            total += group * scale_val
            group = 0
            i += 1
            continue

        # Nếu gặp từ không nhận diện: bỏ qua an toàn
        i += 1

    total += group
    return total


# ---------- Test ngược (một vài ví dụ) ----------
tests = [
    ("không", 0),
    ("một", 1),
    ("mười một", 11),
    ("hai mươi mốt", 21),
    ("ba trăm lẻ bốn", 304),
    ("một nghìn lẻ một", 1001),
    ("một nghìn không trăm lẻ bốn", 1004),
    ("một triệu không trăm lẻ bốn", 1000004),
    ("một tỷ", 1000000000),
    ("một tỷ không trăm lẻ năm", 1000000005),
    ("hai triệu một trăm nghìn không trăm mười", 2100010),
    ("một triệu không trăm lẻ hai nghìn không trăm lẻ ba", 1002003),
    ("một trăm linh hai nghìn một trăm lẻ bốn", 102104),
    ("một trăm lẻ một triệu một trăm mười bốn nghìn không trăm lẻ một", 101114001),
    ("hai tỷ ba trăm bốn mươi lăm triệu sáu trăm bảy mươi tám nghìn chín trăm lẻ một", 2345678901),
]

for text, expect in tests:
    got = vietnamese_to_number(text)
    print(f"{text!r} -> {got}  [{'PASS' if got == expect else 'FAIL, expected '+str(expect)}]")
```

```python
import re
vietnamese_number_words = set([
    "không","một","mốt","hai","ba","bốn","tư","năm","lăm","sáu","bảy","tám","chín",
    "mười","mươi","trăm","nghìn","ngàn","triệu","tỷ","lẻ","linh"
])

def extract_vietnamese_numbers(sentence):
    words = sentence.lower().split()
    result = []
    current = []

    for w in words:
        # Loại bỏ dấu câu đơn giản
        w_clean = re.sub(r"[.,;!?]", "", w)
        if w_clean in vietnamese_number_words:
            current.append(w_clean)
        else:
            if current:
                result.append(" ".join(current))
                current = []
    if current:
        result.append(" ".join(current))
    return result

# Hàm convert từng cụm
def convert_numbers_in_sentence(sentence):
    numbers = extract_vietnamese_numbers(sentence)
    return [vietnamese_to_number(num) for num in numbers]

# ---------- Test ----------
sentence = "Tôi có một trăm lẻ một quả táo và hai nghìn lẻ năm quả cam."
nums_text = extract_vietnamese_numbers(sentence)
nums_value = convert_numbers_in_sentence(sentence)

print("Phát hiện:", nums_text)  # ['một trăm lẻ một', 'hai nghìn lẻ năm']
print("Chuyển thành số:", nums_value)  # [101, 2005]
```

```python
# single_file: entity_detect_and_normalize.py
# -*- coding: utf-8 -*-
"""
One-shot API: process_sentence(sentence)
Returns: (entities, normalized_sentence)
entities: list of dicts {text, case, start, end, normalized}
"""

import re

# --------------------
# (Re-using improved vocab/patterns/parsers from previous code)
NUM_WORDS_SIMPLE = [
    "không","một","mốt","hai","ba","bốn","tư","năm","lăm",
    "sáu","bảy","tám","chín","mười","mươi","trăm","nghìn",
    "triệu","tỷ","linh","lẻ","phẩy"
]

PHONE_DIGIT_WORDS = ["không","một","hai","ba","bốn","năm","sáu","bảy","tám","chín"]

PHONE_WORD_ALT = r"(?:\d+|" + r"|".join(re.escape(w) for w in PHONE_DIGIT_WORDS) + r")"


UNIT_CURRENCY = ["đồng", "đô la", "usd", "vnd"]
UNIT_MEASURE = ["m", "km", "kg", "cm", "mm", "l", "lit"]
ORDINAL_WORDS = {"nhất": 1, "nhì": 2, "thứ nhất": 1, "thứ nhì": 2, "thứ ba": 3, "thứ tư": 4}

NUM_WORD_ALT = r"(?:\d+|" + r"|".join(re.escape(w) for w in NUM_WORDS_SIMPLE) + r")"
NUM_SEQ = r"(?:{nw}(?:\s+{nw})*)".format(nw=NUM_WORD_ALT)

DIGIT_SEQ = r"(?:\d{1,3}(?:[.,]\d{3})*|\d+)"  # cho số 100, 15.800, 1,200,000
NUM_OR_WORD_SEQ = r"(?:" + DIGIT_SEQ + r"|" + NUM_WORD_ALT + r")(?:\s+(?:" + DIGIT_SEQ + r"|" + NUM_WORD_ALT + r"))*"


# Patterns (priority)
_PATTERNS = [
    ("decimal", re.compile(r"\b" + NUM_SEQ + r"\s+phẩy\s+" + NUM_SEQ + r"\b", flags=re.I)),
    ("date", re.compile(r"\bngày\s+" + NUM_SEQ + r"\s+tháng\s+" + NUM_SEQ + r"\s+năm\s+" + NUM_SEQ, flags=re.I)),
    ("date", re.compile(r"\bngày\s+" + NUM_SEQ + r"\s+tháng\s+" + NUM_SEQ, flags=re.I)),
    ("date", re.compile(r"\btháng\s+" + NUM_SEQ + r"\s+năm\s+" + NUM_SEQ, flags=re.I)),
    ("time", re.compile(r"\b" + NUM_SEQ + r"\s+giờ(?:\s+" + NUM_SEQ + r"\s+phút)?(?:\s+" + NUM_SEQ + r"\s+giây)?", flags=re.I)),
    ("phone/account", re.compile(r"\b(?:(?:" + PHONE_WORD_ALT + r")\s+){1,}(?:" + PHONE_WORD_ALT + r")\b", flags=re.I)),
    
    ("currency", re.compile(r"\b" + NUM_OR_WORD_SEQ + r"\s+(?:" + "|".join(re.escape(u) for u in UNIT_CURRENCY) + r")\b", flags=re.I)),
    
    ("percentage", re.compile(r"\b" + NUM_SEQ + r"\s+(?:phần\s+trăm|%)\b", flags=re.I)),
    ("measurement", re.compile(r"\b" + NUM_SEQ + r"\s+(?:" + "|".join(re.escape(u) for u in UNIT_MEASURE) + r")\b", flags=re.I)),
    ("fraction", re.compile(r"\bphần\s+" + NUM_SEQ + r"\b", flags=re.I)),
    ("fraction", re.compile(r"\b" + NUM_SEQ + r"\s+phần\b", flags=re.I)),
    ("ordinal", re.compile(r"\b(?:(?:thứ|hạng)\s+" + NUM_SEQ + r"|\b(?:" + "|".join(re.escape(k) for k in ORDINAL_WORDS.keys()) + r"))\b", flags=re.I)),
    ("number_sequence", re.compile(r"\b" + NUM_SEQ + r"\b", flags=re.I)),
]

# Vietnamese number maps
VI_NUM_MAP = {
    "không":0, "một":1, "mốt":1, "hai":2, "ba":3, "bốn":4, "tư":4,
    "năm":5, "lăm":5, "sáu":6, "bảy":7, "tám":8, "chín":9
}
UNIT_MAP = {"mươi":10, "mười":10, "trăm":100, "nghìn":1000, "triệu":1000000, "tỷ":1000000000}

# overlap helper
def _overlaps(existing_spans, start, end):
    for a,b in existing_spans:
        if not (end <= a or start >= b):
            return True
    return False

# pre-detect: 'cách đây <num> năm'
def predetect_specials(text):
    ents=[]
    for m in re.finditer(r"\bcách đây\s+" + NUM_SEQ + r"\s+năm\b", text, flags=re.I):
        ents.append({"text": m.group().strip(), "case":"year_duration", "start":m.span()[0], "end":m.span()[1]})
    return ents

# merge helper and connector pattern
_NUM_CONNECTORS_RE = re.compile(r'^[\s]*(?:và|,|với|mươi|trăm|nghìn|triệu|tỷ|lẻ|linh|một|hai|ba|bốn|năm|sáu|bảy|tám|chín|mốt|lăm|mười|\d+)[\s]*$', flags=re.I)
def merge_adjacent_entities(entities, text):
    if not entities:
        return []
    merged=[]
    prev=entities[0]
    for cur in entities[1:]:
        gap=text[prev["end"]:cur["start"]]
        if gap.strip()=="" or _NUM_CONNECTORS_RE.match(gap):
            combined_text = text[prev["start"]:cur["end"]].strip()
            priority={"date":7,"time":6,"currency":5,"phone/account":4,"percentage":3,"measurement":3,"ordinal":2,"fraction":2,"number_sequence":1,"decimal":8,"year_duration":9}
            prev_pr=priority.get(prev["case"],0); cur_pr=priority.get(cur["case"],0)
            combined_case = prev["case"] if prev_pr>=cur_pr else cur["case"]
            prev={"text":combined_text,"case":combined_case,"start":prev["start"],"end":cur["end"]}
        else:
            merged.append(prev); prev=cur
    merged.append(prev)
    return merged

# detection core
def detect_number_entities(text):
    if not text:
        return []
    txt=text.strip()
    results=[]; occupied=[]
    for pe in predetect_specials(txt):
        results.append(pe); occupied.append((pe["start"], pe["end"]))
    for case, pattern in _PATTERNS:
        for m in pattern.finditer(txt):
            s,e=m.span()
            if _overlaps(occupied, s, e): continue
            results.append({"text":m.group().strip(), "case":case, "start":s, "end":e})
            occupied.append((s,e))
    results.sort(key=lambda x: x["start"])
    merged = merge_adjacent_entities(results, txt)
    return merged

# normalization helpers
def normalize_decimal(span_text):
    m=re.search(r"(" + NUM_SEQ + r")\s+phẩy\s+(" + NUM_SEQ + r")", span_text, flags=re.I)
    if not m: return span_text
    l,r=m.group(1),m.group(2)
    ln=vietnamese_to_number(l) if l else None
    rn=vietnamese_to_number(r) if r else None
    if ln is None and re.fullmatch(r"(?:\d+\s*)+", l): ln=int(l.replace(" ",""))
    if rn is None and re.fullmatch(r"(?:\d+\s*)+", r): rn=int(r.replace(" ",""))
    if ln is not None and rn is not None: return f"{ln}.{rn}"
    return span_text

def normalize_date(span_text):
    m=re.search(r"ngày\s+(.+?)\s+tháng\s+(.+?)\s+năm\s+(.+)", span_text, flags=re.I)
    if m:
        d_s,mo_s,y_s = m.group(1).strip(), m.group(2).strip(), m.group(3).strip()
        d=vietnamese_to_number(d_s); mo=vietnamese_to_number(mo_s); y=vietnamese_to_number(y_s)
        if d is not None and mo is not None and y is not None:
            return f"{d:02d}/{mo:02d}/{y}"
    m2=re.search(r"ngày\s+(.+?)\s+tháng\s+(.+)", span_text, flags=re.I)
    if m2:
        d_s,mo_s = m2.group(1).strip(), m2.group(2).strip()
        d=vietnamese_to_number(d_s); mo=vietnamese_to_number(mo_s)
        if d is not None and mo is not None: return f"{d:02d}/{mo:02d}"
    return span_text

def normalize_time(span_text):
    m=re.search(r"(" + NUM_SEQ + r")\s+giờ(?:\s+(" + NUM_SEQ + r")\s+phút)?(?:\s+(" + NUM_SEQ + r")\s+giây)?", span_text, flags=re.I)
    if not m: return span_text
    h_s,m_s,s_s = m.group(1), m.group(2), m.group(3)
    h=vietnamese_to_number(h_s) if h_s else None
    mm=vietnamese_to_number(m_s) if m_s else 0
    ss=vietnamese_to_number(s_s) if s_s else None
    if h is not None:
        return f"{h:02d}:{(mm or 0):02d}" + (f":{ss:02d}" if ss is not None else "")
    return span_text

def normalize_percentage(span_text):
    m=re.search(r"(" + NUM_SEQ + r")\s*(?:phần\s+trăm|%)", span_text, flags=re.I)
    if not m: return span_text
    num=vietnamese_to_number(m.group(1))
    if num is not None: return f"{num}%"
    return span_text

def normalize_currency(span_text):
    umatch = re.search(r"(đồng|đô la|usd|vnd)\b", span_text, flags=re.I)
    unit = umatch.group(1) if umatch else ""
    left = re.sub(r"(đồng|đô la|usd|vnd)\b", "", span_text, flags=re.I).strip()

    # Nếu là dạng số có digit (100, 15.800, 1,200,000)
    left_digits = left.replace(".", "").replace(",", "").replace(" ", "")
    if re.fullmatch(r"\d+", left_digits):
        return f"{int(left_digits)} {unit}".strip()

    # Nếu là dạng chữ (một trăm, hai mươi nghìn, ...)
    num = vietnamese_to_number(left)
    if num is not None:
        return f"{num} {unit}".strip()

    return span_text


def normalize_phone(span_text):
    tokens = span_text.lower().strip().split()
    digits = []
    for t in tokens:
        if re.fullmatch(r"\d+", t):  # số đã viết liền (vd: 545433)
            digits.extend(list(t))   # tách thành từng digit
        elif t in VI_NUM_MAP:        # số đọc bằng chữ
            digits.append(str(VI_NUM_MAP[t]))
        # bỏ qua từ không hợp lệ
    return "".join(digits) if digits else span_text






def normalize_number_sequence(span_text):
    s = span_text.lower().strip()

    # --- đặc biệt: "<num> linh|lẻ <num>" ---
    m = re.fullmatch(r"(\w+)\s+(linh|lẻ)\s+(\w+)", s)
    if m:
        left, _, right = m.groups()
        left_num = vietnamese_to_number(left)
        right_num = vietnamese_to_number(right)
        if left_num is not None and right_num is not None:
            return str(left_num * 100 + right_num)

    # --- nếu toàn là số (có thể có khoảng trắng) ---
    if re.fullmatch(r"(?:\d+\s*)+", s):
        return "".join(s.split())

    # --- chuyển từ chữ sang số ---
    num = vietnamese_to_number(s)
    if num is not None:
        return str(num)

    return span_text


def normalize_fraction(span_text):
    m=re.search(r"phần\s+(" + NUM_SEQ + r")", span_text, flags=re.I)
    if m:
        denom=vietnamese_to_number(m.group(1))
        if denom: return f"1/{denom}"
    m2=re.search(r"(" + NUM_SEQ + r")\s+phần", span_text, flags=re.I)
    if m2:
        num=vietnamese_to_number(m2.group(1))
        if num: return f"{num}/?"
    return span_text

def normalize_ordinal(span_text):
    s=span_text.lower().strip()
    if s in ORDINAL_WORDS: return str(ORDINAL_WORDS[s])
    m=re.search(r"(?:hạng|thứ)\s+(.+)", s)
    if m:
        k=m.group(1).strip()
        if k in ORDINAL_WORDS: return f"{m.group(0).split()[0]} {ORDINAL_WORDS[k]}"
        nv=vietnamese_to_number(k)
        if nv is not None: return f"{m.group(0).split()[0]} {nv}"
    return span_text

# normalization dispatcher
_NORMALIZER = {
    "date": normalize_date,
    "time": normalize_time,
    "currency": normalize_currency,
    "percentage": normalize_percentage,
    "phone/account": normalize_phone,
    "number_sequence": normalize_number_sequence,
    "decimal": normalize_decimal,
    "fraction": normalize_fraction,
    "ordinal": normalize_ordinal,
    "year_duration": lambda s: re.sub(r"(cách đây\s+)(" + NUM_SEQ + r")(\s+năm)", lambda m: m.group(1) + (str(vietnamese_words_to_int(m.group(2)) if vietnamese_words_to_int(m.group(2)) is not None else m.group(2))) + m.group(3), s, flags=re.I)
}


def process_sentence(sentence):
    ents = detect_number_entities(sentence)
    enriched = []

    for e in ents:
        case = e["case"]
        text_span = e["text"]

        norm_fn = _NORMALIZER.get(case)
        if norm_fn:
            try:
                norm = norm_fn(text_span)
            except Exception:
                norm = text_span
        else:
            tmp = vietnamese_to_number(text_span)
            norm = str(tmp) if tmp is not None else text_span

        enriched.append({
            "text": text_span,
            "case": case,
            "start": e["start"],
            "end": e["end"],
            "normalized": norm,
        })

    return enriched, normalize_detected_entities(sentence, enriched)



# reuse detect & normalize functions from above
def detect_number_entities(text):
    return detect_number_entities__impl(text)

# To avoid recursion, move actual implementation under a different name:
def detect_number_entities__impl(text):
    if not text: return []
    txt=text.strip(); results=[]; occupied=[]
    for pe in predetect_specials(txt):
        results.append(pe); occupied.append((pe["start"], pe["end"]))
    for case, pattern in _PATTERNS:
        for m in pattern.finditer(txt):
            s,e=m.span()
            if _overlaps(occupied,s,e): continue
            results.append({"text":m.group().strip(), "case":case, "start":s, "end":e})
            occupied.append((s,e))
    results.sort(key=lambda x: x["start"])
    return merge_adjacent_entities(results, txt)

# normalize function must reuse the earlier normalizers; define wrapper here
def normalize_detected_entities(text, detected_entities):
    if not detected_entities:
        return text
    txt = text
    ents = sorted(detected_entities, key=lambda x: x["start"])
    out_parts=[]; last=0
    for ent in ents:
        s,e=ent["start"], ent["end"]
        out_parts.append(txt[last:s])
        span=ent["text"]; case=ent["case"]
        norm = _NORMALIZER.get(case, lambda s: s)(span) if case in _NORMALIZER else (str(vietnamese_words_to_int(span)) if vietnamese_words_to_int(span) is not None else span)
        out_parts.append(norm)
        last=e
    out_parts.append(txt[last:])
    return "".join(out_parts)

# # --------------------
# # quick self-test if run as script
# if __name__ == "__main__":
#     samples = [
#         "gọi một chín tám",
#         "cách đây mười hai năm đồng đô la có giá khoảng mười lăm nghìn tám trăm đồng",
#         "lãi suất là mười hai phần trăm",
#         "sinh nhật vào ngày mười hai tháng ba năm hai nghìn hai mươi lăm",
#         "số điện thoại là không một hai ba bốn năm sáu bảy tám chín",
#         "hạng nhất thuộc về đội A",
#         "cuộc họp bắt đầu vào mười giờ ba mươi phút",
#         "giá là 15 800 đồng",
#         "có 1 phẩy 5 điểm",
#         "tổng cộng là hai mươi mốt nghìn đồng",
#         "anh ta nói mười lăm",
#     ]
#     for s in samples:
#         ents, norm = process_sentence(s)
#         print("Input:", s)
#         print("Entities:")
#         for ent in ents:
#             print(" ", ent)
#         print("Normalized:", norm)
#         print()
```

```python
# test_runner_fixed.py
# -*- coding: utf-8 -*-
"""
Fixed test runner: handles test entries that are dict, str, tuple/list, or other simple forms.
Expects process_sentence to be importable or defined in the same namespace.
"""

import re
import importlib
import sys
from collections import defaultdict

# ---------------------------
def find_process_sentence():
    candidates = [
        ("entity_detect_and_normalize", "process_sentence"),
        ("process_sentence_module", "process_sentence"),
        ("__main__", "process_sentence"),
    ]
    for mod_name, fn in candidates:
        try:
            if mod_name == "__main__":
                if fn in globals():
                    return globals()[fn]
            else:
                m = importlib.import_module(mod_name)
                if hasattr(m, fn):
                    return getattr(m, fn)
        except Exception:
            continue
    if "process_sentence" in globals():
        return globals()["process_sentence"]
    raise RuntimeError("process_sentence not found. Ensure the processing module is importable.")

# ---------------------------
def _normalize_compare(s):
    if s is None:
        return None
    s = str(s).lower().strip()
    s = re.sub(r"[–—]", "-", s)
    s = re.sub(r"\s+", " ", s)
    return s

def expected_matches(expected, normalized_sentence, entities_normalized_texts):
    if expected is None:
        return None  # SKIP/INFO

    if isinstance(expected, (tuple, list)):
        for item in expected:
            if _normalize_compare(str(item)) not in _normalize_compare(normalized_sentence):
                return False
        return True

    e = _normalize_compare(str(expected))
    ns = _normalize_compare(normalized_sentence or "")
    if e in ns:
        return True
    for ent_txt in entities_normalized_texts:
        if _normalize_compare(ent_txt) == e:
            return True
        if e in _normalize_compare(ent_txt or ""):
            return True

    def digits_only(x):
        return re.sub(r"[^0-9\.\-%]", "", x or "")
    if digits_only(e) and digits_only(e) in digits_only(ns):
        return True

    return False

# ---------------------------
TEST_LIST_NAMES = [
    "TEST_INTEGER", "TEST_DECIMAL", "TEST_PERCENTAGE", "TEST_CURRENCY",
    "TEST_MEASUREMENT", "TEST_DATE", "TEST_TIME", "TEST_PHONE",
    "TEST_ORDINAL", "TEST_FRACTION", "TEST_NUMBER_SEQUENCE",
    "TEST_YEAR_DURATION", "TEST_EDGE"
]

def collect_tests():
    tests = {}
    # Prefer global vars in this module
    for name in TEST_LIST_NAMES:
        if name in globals():
            tests[name] = globals()[name]
    # add any other TEST_* lists present
    for k, v in globals().items():
        if k.startswith("TEST_") and k not in tests and isinstance(v, list):
            tests[k] = v
    if not tests:
        raise RuntimeError("No TEST_* lists found in globals. Paste the test lists into this script or import them.")
    return tests

# ---------------------------
def _normalize_test_entry(t):
    """
    Accepts multiple test entry formats:
      - dict with keys 'input' and optional 'expected'
      - str: treated as {"input": str, "expected": None}
      - tuple/list of (input, expected)
      - simple other -> str(input)
    Returns (input_str, expected)
    """
    if isinstance(t, dict):
        return t.get("input"), t.get("expected")
    if isinstance(t, str):
        return t, None
    if isinstance(t, (list, tuple)):
        if len(t) == 0:
            return None, None
        if len(t) == 1:
            return t[0], None
        return t[0], t[1]
    # fallback: convert to string
    return str(t), None

def run_all_tests(process_fn, tests_map):
    summary = {}
    detailed = defaultdict(list)

    total = 0
    passed = 0
    skipped = 0

    for group_name, test_list in tests_map.items():
        group_tot = 0
        group_pass = 0
        group_skip = 0
        for raw_t in test_list:
            group_tot += 1
            total += 1
            inp, expected = _normalize_test_entry(raw_t)
            if inp is None:
                detailed[group_name].append({"input": raw_t, "expected": expected, "error": "Empty test entry"})
                continue
            try:
                ents, normalized = process_fn(inp)
            except Exception as exc:
                detailed[group_name].append({
                    "input": inp, "expected": expected, "error": str(exc)
                })
                continue

            ents_norm_texts = []
            if isinstance(ents, list):
                for ent in ents:
                    if isinstance(ent, dict) and "normalized" in ent:
                        ents_norm_texts.append(ent["normalized"])
                    elif isinstance(ent, dict) and "text" in ent:
                        ents_norm_texts.append(ent.get("normalized") or ent.get("text"))
                    else:
                        ents_norm_texts.append(str(ent))
            else:
                # if ents is not list, just stringify
                ents_norm_texts.append(str(ents))

            match = expected_matches(expected, normalized, ents_norm_texts)
            if match is None:
                status = "SKIP"
                group_skip += 1
                skipped += 1
            elif match is True:
                status = "PASS"
                group_pass += 1
                passed += 1
            else:
                status = "FAIL"

            detailed[group_name].append({
                "input": inp,
                "expected": expected,
                "normalized": normalized,
                "entities": ents,
                "result": status
            })
        summary[group_name] = {"total": group_tot, "passed": group_pass, "skipped": group_skip, "failed": group_tot - group_pass - group_skip}
    return {
        "summary": summary,
        "details": detailed,
        "total": total,
        "passed": passed,
        "skipped": skipped,
        "failed": total - passed - skipped
    }

# ---------------------------
def pretty_print_report(report):
    print("\nTEST SUMMARY")
    print("-----------")
    for grp, stats in report["summary"].items():
        print(f"{grp}: total={stats['total']}  passed={stats['passed']}  skipped={stats['skipped']}  failed={stats['failed']}")
    print("-----------")
    print(f"TOTAL: {report['total']}, PASSED: {report['passed']}, SKIPPED: {report['skipped']}, FAILED: {report['failed']}")
    print("\nDETAILS (failures & examples):\n")
    for grp, entries in report["details"].items():
        print(f"== {grp} ==")
        for e in entries:
            if e.get("result") == "FAIL" or e.get("result") == "SKIP" or "error" in e:
                print("INPUT:", e["input"])
                print(" EXPECTED:", e.get("expected"))
                if "error" in e:
                    print(" ERROR:", e["error"])
                else:
                    print(" NORMALIZED:", e.get("normalized"))
                    print(" ENTITIES:", e.get("entities"))
                    print(" RESULT:", e.get("result"))
                print("-")
        print()

# ---------------------------
if __name__ == "__main__":
    try:
        process_fn = find_process_sentence()
    except Exception as exc:
        print("Error: cannot find process_sentence function:", exc)
        sys.exit(2)

    try:
        tests_map = collect_tests()
    except Exception as exc:
        print("Error: cannot find test lists in globals.", exc)
        sys.exit(2)

    report = run_all_tests(process_fn, tests_map)
    pretty_print_report(report)

    # if report["failed"] > 0:
    #     sys.exit(1)
    # else:
    #     sys.exit(0)
```
