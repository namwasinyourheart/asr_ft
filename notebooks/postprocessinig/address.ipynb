```python
import re

def replace_tren_with_slash(text: str) -> str:
    """
    Detect and convert number sequences containing 'Trên' into slash-separated format.
    
    Example:
        "15 Trên 6 Trên 89" -> "15/6/89"
        "Nhà tôi nằm Trên đường lớn, số 15 Trên 6 Trên 89"
            -> "Nhà tôi nằm Trên đường lớn, số 15/6/89"
    
    Notes:
        - Only affects cases where 'Trên' is between digits.
        - Keeps all other 'Trên' (not part of a number sequence) unchanged.
    """
    pattern = re.compile(r'\d+(?:\s+(?:Tr[eê]n)\s+\d+)+', flags=re.IGNORECASE)

    def repl(match: re.Match) -> str:
        seq = match.group(0)
        return re.sub(r'\s*(?:Tr[eê]n)\s*', '/', seq, flags=re.IGNORECASE)

    return pattern.sub(repl, text)


# --- Quick test ---
examples = [
    "15 Trên 6 Trên 89 Tô Ngọc Vân, Quận 12",
    "Nhà tôi nằm Trên đường lớn, số 15 Trên 6 Trên 89 Quận 12",
    "7 Trên 2 Nguyễn Thái Học, Quận 1",
    "Không có chữ Trên trong số liệu này",
    "5 Trên 3 Trên 2 Trên 1 Khu phố 9",
    "15 Trên 6 Trên 8 và 22 Trên 11 Trên 5 Đường XYZ",
]

for ex in examples:
    print(ex)
    print(" ->", replace_tren_with_slash(ex))
```
