```python
PROPER_NOUNS = {"hà nội", "hồ chí minh", "vietnam", "fpt"}

def capitalize_proper_nouns(text):
    words = text.split()
    i = 0
    output = []
    while i < len(words):
        # thử ghép bigram/trigram để match multi-word proper noun
        for n in (3,2,1):
            phrase = " ".join(words[i:i+n])
            if phrase.lower() in PROPER_NOUNS:
                output.append(phrase.title())  # viết hoa mỗi từ
                i += n
                break
        else:
            output.append(words[i])
            i += 1
    return " ".join(output)
```

```python
from rapidfuzz import process, fuzz

def correct_locations_3level(ner_results, wards=None, districts=None, provinces=None, threshold=80):
    """
    ner_results: output từ pipeline NER (list of dict)
    wards: list tên xã/phường chuẩn
    districts: list tên quận/huyện chuẩn
    provinces: list tên tỉnh/thành chuẩn
    threshold: minimum similarity score (%)
    
    Trả về list tuple: (original_text, corrected_text, level, score)
    """
    corrected = []
    
    for ent in ner_results:
        if ent.get("entity_group") in ["LOC", "LOCATION"]:
            loc = ent["word"]
            best_match = None
            best_score = 0
            level = None
            
            # ưu tiên cấp nhỏ nhất trước
            for name_list, lvl in zip([wards, districts, provinces], ['ward','district','province']):
                if name_list:
                    match, score, _ = process.extractOne(loc, name_list, scorer=fuzz.WRatio)
                    if score > best_score:
                        best_score = score
                        best_match = match
                        level = lvl
            
            if best_score >= threshold:
                corrected.append((loc, best_match, level, best_score))
            else:
                corrected.append((loc, None, None, best_score))
    
    return corrected

# Ví dụ danh sách chuẩn
wards = ["Phường 1", "Phường Tân Phú", "Xã Hữu Lũng", "Xã Phước Thắng"]
districts = ["Quận 1", "Quận 3", "Huyện Hữu Lũng", "Huyện Ninh Phước"]
provinces = ["Hà Nội", "Hồ Chí Minh", "Đà Nẵng", "Ninh Thuận", "Bình Thuận"]

# Giả sử ner_results từ pipeline NER
ner_results = [
    {"word": "tỉnh Ninh Thuậtttt", "entity_group": "LOCATION"},
    {"word": "Huyện Ninh Phướcttt", "entity_group": "LOCATION"},
    {"word": "Xã Phước Thắk", "entity_group": "LOCATION"}
]

results = correct_locations_3level(ner_results, wards, districts, provinces)
for orig, corr, lvl, score in results:
    print(f"{orig} → {corr} (level={lvl}, score={score})")
```

```python
from transformers import AutoTokenizer, AutoModelForTokenClassification, pipeline

# Load model once
tokenizer = AutoTokenizer.from_pretrained("NlpHUST/ner-vietnamese-electra-base")
model = AutoModelForTokenClassification.from_pretrained("NlpHUST/ner-vietnamese-electra-base")
nlp_ner = pipeline("ner", model=model, tokenizer=tokenizer)

def capitalize_entity(text: str) -> str:
    """Viết hoa chữ cái đầu của một cụm từ, giữ nguyên các từ khác."""
    return ' '.join(w.capitalize() for w in text.split())

def postprocess_uppercase(text: str) -> str:
    """
    Nhận input text, thực hiện NER, viết hoa chữ cái đầu các entity: PERSON và LOCATION.
    Trả về text đã được postprocess.
    """
    ner_results = nlp_ner(text)

    if not ner_results:
        return text

    # Gộp các token liên tiếp cùng entity (B/I)
    merged_entities = []
    current = None
    for ent in ner_results:
        if ent['entity'].startswith("B-"):
            if current:
                merged_entities.append(current)
            current = {
                "start": ent['start'],
                "end": ent['end'],
                "entity": ent['entity'][2:],
                "text": ent['word']
            }
        elif ent['entity'].startswith("I-") and current and current['entity'] == ent['entity'][2:]:
            # nối token tiếp
            current['end'] = ent['end']
            current['text'] += ' ' + ent['word']
        else:
            if current:
                merged_entities.append(current)
            current = None
    if current:
        merged_entities.append(current)

    # Tạo text postprocess
    out_parts = []
    last_idx = 0
    for ent in merged_entities:
        out_parts.append(text[last_idx:ent['start']])
        if ent['entity'] in ('PERSON', 'LOCATION', 'ORGANIZATION'):
            out_parts.append(capitalize_entity(ent['text']))
        else:
            out_parts.append(ent['text'])
        last_idx = ent['end']
    out_parts.append(text[last_idx:])

    return ''.join(out_parts)
```

```python
from rapidfuzz import process, fuzz

known_orgs = ["Liên Việt Postbank", "Vietcombank", "Techcombank"]

def correct_org(asr_text):
    # extractOne trả về một dict-like object
    match = process.extractOne(asr_text, known_orgs, scorer=fuzz.ratio)
    if match and match[1] > 70:  # match[0]=best string, match[1]=score
        return match[0]
    return asr_text

text = "điên việt vốt spanh vừa khai trương chi nhánh mới"
print(correct_org(text))
# -> "Liên Việt Postbank"
```
